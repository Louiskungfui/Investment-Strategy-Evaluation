```python
# Investment Portfolio Project
# Step 1: Import market data
#
# Assets:
# SPY = US stocks (S&P 500 ETF)
# IEF = US government bonds (7-10 Year Treasury ETF)
# BIL = Cash proxy (1-3 Month US Treasury Bill ETF)
#
# The backtest will begin in January 2015.
# We download data from March 2014 so that we have
# 10 months of previous data for the moving average calculation.

# If yfinance is not installed, run this first in Google Colab:
# !pip install yfinance

import yfinance as yf
import pandas as pd

# 1. Set the dates

start_date = "2014-01-01"
end_date = "2026-01-01"   # yfinance end date is exclusive

# 2. Set the assets

tickers = ["SPY", "IEF", "BIL"]

# 3. Download daily price data

data = yf.download(
    tickers,
    start=start_date,
    end=end_date,
    auto_adjust=False,
    progress=False
)

# 4. Keep adjusted close prices

# Adjusted Close is used because it accounts for dividends
# and other distributions.

prices = data["Adj Close"].copy()

# 5. Check the data

print("First 5 rows:")
print(prices.head())

print("\nLast 5 rows:")
print(prices.tail())

print("\nMissing values:")
print(prices.isna().sum())

print("\nDate range:")
print(prices.index.min(), "to", prices.index.max())
```

```python
#6. calcuating moving average

# Convert daily adjusted prices into month-end prices
monthly_prices = prices.resample("ME").last()

# Function to calculate a moving average of any chosen length
def calculate_moving_average(months):
    return monthly_prices[["SPY", "IEF"]].rolling(window=months).mean()

# Example: 10-month moving average
moving_average = calculate_moving_average(10)


print(moving_average.head(15))
```

```python
#7. Calucating returns

# Calculate monthly returns for all three assets
monthly_returns = monthly_prices.pct_change()

# Create moving average signals
# 1 = asset is above its moving average
# 0 = asset is below its moving average
signals = (monthly_prices[["SPY", "IEF"]] > moving_average).astype(int)

# Shift the signals forward by one month
# This means the signal at the end of one month determines
# what we hold during the following month
signals = signals.shift(1)

# Keep only the backtest period
monthly_returns = monthly_returns.loc["2015-01-01":"2025-12-31"]
signals = signals.loc["2015-01-01":"2025-12-31"]

print("Monthly returns:")
print(monthly_returns.head())

print("\nTrading signals:")
print(signals.head())
```

```python
#8. investment strategy 1.

# Strategy 1: 60/40 stocks and bonds, rebalanced monthly

stock_weight = 0.60
bond_weight = 0.40

strategy_1_returns = (
    stock_weight * monthly_returns["SPY"]
    + bond_weight * monthly_returns["IEF"]
)

# Start with a theoretical $100,000 portfolio
starting_value = 100000

strategy_1_value = starting_value * (1 + strategy_1_returns).cumprod()

print("Strategy 1 monthly returns:")
print(strategy_1_returns.head())

print("\nStrategy 1 portfolio value:")
print(strategy_1_value.head())
```

```python
#9. Investment strategy 2.

# Strategy 2: 60/40 portfolio with moving-average signals

# Set the portfolio weights for each month
strategy_2_weights = pd.DataFrame(index=signals.index)

strategy_2_weights["SPY"] = 0.60 * signals["SPY"]
strategy_2_weights["IEF"] = 0.40 * signals["IEF"]

# Any allocation removed from stocks or bonds is moved into cash
strategy_2_weights["BIL"] = (
    1
    - strategy_2_weights["SPY"]
    - strategy_2_weights["IEF"]
)

# Calculate the monthly portfolio return
strategy_2_returns = (
    strategy_2_weights["SPY"] * monthly_returns["SPY"]
    + strategy_2_weights["IEF"] * monthly_returns["IEF"]
    + strategy_2_weights["BIL"] * monthly_returns["BIL"]
)

# Track the value of a theoretical $100,000 portfolio
strategy_2_value = starting_value * (1 + strategy_2_returns).cumprod()

print("Strategy 2 weights:")
print(strategy_2_weights.head())

print("\nStrategy 2 monthly returns:")
print(strategy_2_returns.head())

print("\nStrategy 2 portfolio value:")
print(strategy_2_value.head())
```

```python
# Create an S&P 500 benchmark starting at $100,000
spy_value = starting_value * (1 + monthly_returns["SPY"]).cumprod()

# Put all portfolio values into one table
portfolio_values = pd.DataFrame({
    "60/40 Portfolio": strategy_1_value,
    "Moving Average Portfolio": strategy_2_value,
    "S&P 500": spy_value
})

print(portfolio_values)

# Plot portfolio growth
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.plot(portfolio_values.index,
         portfolio_values["60/40 Portfolio"],
         label="60/40 Portfolio")

plt.plot(portfolio_values.index,
         portfolio_values["Moving Average Portfolio"],
         label="Moving Average Portfolio")

plt.plot(portfolio_values.index,
         portfolio_values["S&P 500"],
         label="S&P 500")

plt.title("Growth of $100,000: January 2015 - December 2025")
plt.xlabel("Date")
plt.ylabel("Portfolio Value ($)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Put the monthly returns into one table
portfolio_returns = pd.DataFrame({
    "60/40 Portfolio": strategy_1_returns,
    "Moving Average Portfolio": strategy_2_returns,
    "S&P 500": monthly_returns["SPY"]
})

# Function to calculate the main performance metrics
def calculate_metrics(returns, values):
    years = len(returns) / 12

    annualised_return = (values.iloc[-1] / starting_value) ** (1 / years) - 1

    annualised_volatility = returns.std() * (12 ** 0.5)

    running_max = values.cummax()
    drawdown = values / running_max - 1
    max_drawdown = drawdown.min()

    excess_returns = returns - monthly_returns["BIL"]
    sharpe_ratio = (
        excess_returns.mean()
        / excess_returns.std()
        * (12 ** 0.5)
    )

    negative_returns = returns[returns < 0]
    downside_deviation = (
        (negative_returns ** 2).mean() ** 0.5
    )

    sortino_ratio = (
        returns.mean()
        / downside_deviation
        * (12 ** 0.5)
    )

    return {
        "Annualised Return": annualised_return,
        "Annualised Volatility": annualised_volatility,
        "Maximum Drawdown": max_drawdown,
        "Sharpe Ratio": sharpe_ratio,
        "Sortino Ratio": sortino_ratio,
        "Final Value": values.iloc[-1]
    }

# Calculate metrics for each portfolio
results = pd.DataFrame({
    "60/40 Portfolio": calculate_metrics(
        strategy_1_returns,
        strategy_1_value
    ),

    "Moving Average Portfolio": calculate_metrics(
        strategy_2_returns,
        strategy_2_value
    ),

    "S&P 500": calculate_metrics(
        monthly_returns["SPY"],
        spy_value
    )
}).T

# Format percentage columns
percentage_columns = [
    "Annualised Return",
    "Annualised Volatility",
    "Maximum Drawdown"
]

for column in percentage_columns:
    results[column] = results[column] * 100

print(results.round(2))
```

```python
# Calculate calendar year returns
annual_returns = (1 + portfolio_returns).resample("YE").prod() - 1

# Best and worst calendar years
best_year = annual_returns.max()
worst_year = annual_returns.min()

# Percentage of months with positive returns
positive_months = (portfolio_returns > 0).mean()

# Average monthly gain and loss
average_positive_month = portfolio_returns.where(
    portfolio_returns > 0
).mean()

average_negative_month = portfolio_returns.where(
    portfolio_returns < 0
).mean()

# Create a client-focused summary table
client_metrics = pd.DataFrame({
    "Best Year": best_year * 100,
    "Worst Year": worst_year * 100,
    "Positive Months": positive_months * 100,
    "Average Positive Month": average_positive_month * 100,
    "Average Negative Month": average_negative_month * 100
})

print("Client-focused performance metrics:")
print(client_metrics.round(2))

print("\nCalendar year returns:")
print((annual_returns * 100).round(2))
```

```python
# Calculate drawdowns for each portfolio
drawdowns = portfolio_values / portfolio_values.cummax() - 1

# Function to analyse the worst drawdown
def drawdown_analysis(values):
    running_max = values.cummax()
    drawdown = values / running_max - 1

    # Date of the worst point
    trough_date = drawdown.idxmin()

    # Previous portfolio peak before the drawdown
    peak_date = values.loc[:trough_date].idxmax()
    peak_value = values.loc[peak_date]

    # Find when the portfolio recovers back to the previous peak
    after_trough = values.loc[trough_date:]
    recovery = after_trough[after_trough >= peak_value]

    if len(recovery) > 0:
        recovery_date = recovery.index[0]

        peak_position = values.index.get_loc(peak_date)
        trough_position = values.index.get_loc(trough_date)
        recovery_position = values.index.get_loc(recovery_date)

        months_to_bottom = trough_position - peak_position
        months_to_recover = recovery_position - trough_position
        total_drawdown_months = recovery_position - peak_position

    else:
        recovery_date = "Not recovered"
        months_to_bottom = values.index.get_loc(trough_date) - values.index.get_loc(peak_date)
        months_to_recover = None
        total_drawdown_months = None

    return {
        "Peak Date": peak_date,
        "Trough Date": trough_date,
        "Maximum Drawdown (%)": drawdown.min() * 100,
        "Recovery Date": recovery_date,
        "Months to Bottom": months_to_bottom,
        "Months to Recover": months_to_recover,
        "Total Drawdown Months": total_drawdown_months
    }

# Analyse all three portfolios
drawdown_results = pd.DataFrame({
    "60/40 Portfolio": drawdown_analysis(
        portfolio_values["60/40 Portfolio"]
    ),

    "Moving Average Portfolio": drawdown_analysis(
        portfolio_values["Moving Average Portfolio"]
    ),

    "S&P 500": drawdown_analysis(
        portfolio_values["S&P 500"]
    )
}).T

print("Worst drawdown analysis:")
print(drawdown_results)

# Plot drawdowns over time
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.plot(
    drawdowns.index,
    drawdowns["60/40 Portfolio"],
    label="60/40 Portfolio"
)

plt.plot(
    drawdowns.index,
    drawdowns["Moving Average Portfolio"],
    label="Moving Average Portfolio"
)

plt.plot(
    drawdowns.index,
    drawdowns["S&P 500"],
    label="S&P 500"
)

plt.axhline(0)

plt.title("Portfolio Drawdowns: 2015 - 2025")
plt.xlabel("Date")
plt.ylabel("Drawdown")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Robustness test for different moving-average periods

def test_moving_average_strategy(ma_months):
    moving_average_test = (
        monthly_prices[["SPY", "IEF"]]
        .rolling(window=ma_months)
        .mean()
    )

    signals_test = (
        monthly_prices[["SPY", "IEF"]] > moving_average_test
    ).astype(int)

    signals_test = signals_test.shift(1)
    signals_test = signals_test.loc["2015-01-01":"2025-12-31"]

    weights = pd.DataFrame(index=signals_test.index)

    weights["SPY"] = 0.60 * signals_test["SPY"]
    weights["IEF"] = 0.40 * signals_test["IEF"]
    weights["BIL"] = 1 - weights["SPY"] - weights["IEF"]

    returns = (
        weights["SPY"] * monthly_returns["SPY"]
        + weights["IEF"] * monthly_returns["IEF"]
        + weights["BIL"] * monthly_returns["BIL"]
    )

    values = starting_value * (1 + returns).cumprod()

    years = len(returns) / 12

    annualised_return = (
        values.iloc[-1] / starting_value
    ) ** (1 / years) - 1

    annualised_volatility = returns.std() * (12 ** 0.5)

    running_max = values.cummax()
    drawdown = values / running_max - 1
    max_drawdown = drawdown.min()

    excess_returns = returns - monthly_returns["BIL"]

    sharpe_ratio = (
        excess_returns.mean()
        / excess_returns.std()
        * (12 ** 0.5)
    )

    return {
        "Annualised Return (%)": annualised_return * 100,
        "Annualised Volatility (%)": annualised_volatility * 100,
        "Maximum Drawdown (%)": max_drawdown * 100,
        "Sharpe Ratio": sharpe_ratio,
        "Final Value ($)": values.iloc[-1]
    }


robustness_results = pd.DataFrame({
    "6 Month MA": test_moving_average_strategy(6),
    "8 Month MA": test_moving_average_strategy(8),
    "10 Month MA": test_moving_average_strategy(10),
    "12 Month MA": test_moving_average_strategy(12)
}).T


# Add the conventional 60/40 portfolio for comparison

years = len(strategy_1_returns) / 12

strategy_1_annualised_return = (
    strategy_1_value.iloc[-1] / starting_value
) ** (1 / years) - 1

strategy_1_volatility = strategy_1_returns.std() * (12 ** 0.5)

strategy_1_drawdown = (
    strategy_1_value / strategy_1_value.cummax() - 1
)

strategy_1_excess_returns = (
    strategy_1_returns - monthly_returns["BIL"]
)

strategy_1_sharpe = (
    strategy_1_excess_returns.mean()
    / strategy_1_excess_returns.std()
    * (12 ** 0.5)
)

robustness_results.loc["60/40 Portfolio"] = {
    "Annualised Return (%)": strategy_1_annualised_return * 100,
    "Annualised Volatility (%)": strategy_1_volatility * 100,
    "Maximum Drawdown (%)": strategy_1_drawdown.min() * 100,
    "Sharpe Ratio": strategy_1_sharpe,
    "Final Value ($)": strategy_1_value.iloc[-1]
}

print("Moving-average robustness test:")
print(robustness_results.round(2))
```

```python
# Compare how the different moving-average rules behave

def analyse_ma_behaviour(ma_months):

    ma = (
        monthly_prices[["SPY", "IEF"]]
        .rolling(window=ma_months)
        .mean()
    )

    test_signals = (
        monthly_prices[["SPY", "IEF"]] > ma
    ).astype(int)

    test_signals = test_signals.shift(1)
    test_signals = test_signals.loc["2015-01-01":"2025-12-31"]

    weights = pd.DataFrame(index=test_signals.index)

    weights["SPY"] = 0.60 * test_signals["SPY"]
    weights["IEF"] = 0.40 * test_signals["IEF"]
    weights["BIL"] = 1 - weights["SPY"] - weights["IEF"]

    allocation_changes = (
        weights.diff().abs().sum(axis=1) > 0
    ).sum()

    fully_invested = (
        (weights["SPY"] == 0.60)
        & (weights["IEF"] == 0.40)
    ).sum()

    stock_defensive = (
        (weights["SPY"] == 0)
        & (weights["IEF"] == 0.40)
    ).sum()

    bond_defensive = (
        (weights["SPY"] == 0.60)
        & (weights["IEF"] == 0)
    ).sum()

    fully_defensive = (
        weights["BIL"] == 1
    ).sum()

    average_cash = weights["BIL"].mean()

    return {
        "Fully Invested Months": fully_invested,
        "Equity Defensive Months": stock_defensive,
        "Bond Defensive Months": bond_defensive,
        "Fully Defensive Months": fully_defensive,
        "Allocation Changes": allocation_changes,
        "Average Cash Allocation (%)": average_cash * 100
    }


ma_behaviour = pd.DataFrame({
    "6 Month": analyse_ma_behaviour(6),
    "8 Month": analyse_ma_behaviour(8),
    "10 Month": analyse_ma_behaviour(10),
    "12 Month": analyse_ma_behaviour(12)
}).T

print(ma_behaviour.round(2))
```

```python
# Transaction cost analysis
# Base assumption: 5 basis points for each dollar bought or sold

transaction_cost_bps = 5
transaction_cost_rate = transaction_cost_bps / 10000

def create_ma_weights(ma_months):
    moving_average_test = (
        monthly_prices[["SPY", "IEF"]]
        .rolling(window=ma_months)
        .mean()
    )

    signals_test = (
        monthly_prices[["SPY", "IEF"]] > moving_average_test
    ).astype(int)

    signals_test = signals_test.shift(1)
    signals_test = signals_test.loc["2015-01-01":"2025-12-31"]

    weights = pd.DataFrame(index=signals_test.index)

    weights["SPY"] = 0.60 * signals_test["SPY"]
    weights["IEF"] = 0.40 * signals_test["IEF"]
    weights["BIL"] = 1 - weights["SPY"] - weights["IEF"]

    return weights


def backtest_with_transaction_costs(target_weights, cost_bps=5):
    cost_rate = cost_bps / 10000

    portfolio_value = starting_value
    previous_value = starting_value

    net_returns = []
    portfolio_values_net = []
    one_way_turnover = []
    transaction_costs_dollars = []

    current_weights = target_weights.iloc[0].copy()

    for i, date in enumerate(target_weights.index):

        target = target_weights.loc[date]

        if i == 0:
            turnover = 0
            transaction_cost = 0

        else:
            weight_changes = target - current_weights

            # One-way turnover avoids double-counting buys and sells
            turnover = weight_changes.abs().sum() / 2

            # Costs are charged on both purchases and sales
            gross_traded_notional = weight_changes.abs().sum()

            transaction_cost = (
                portfolio_value
                * gross_traded_notional
                * cost_rate
            )

            portfolio_value -= transaction_cost

        current_weights = target.copy()

        asset_returns = monthly_returns.loc[
            date, ["SPY", "IEF", "BIL"]
        ]

        portfolio_return = (
            current_weights * asset_returns
        ).sum()

        portfolio_value *= (1 + portfolio_return)

        # Allow the weights to drift during the month
        end_values = (
            current_weights
            * (1 + asset_returns)
        )

        current_weights = end_values / end_values.sum()

        net_monthly_return = (
            portfolio_value / previous_value - 1
        )

        net_returns.append(net_monthly_return)
        portfolio_values_net.append(portfolio_value)
        one_way_turnover.append(turnover)
        transaction_costs_dollars.append(transaction_cost)

        previous_value = portfolio_value

    results = pd.DataFrame(
        {
            "Net Return": net_returns,
            "Portfolio Value": portfolio_values_net,
            "One-Way Turnover": one_way_turnover,
            "Transaction Cost ($)": transaction_costs_dollars
        },
        index=target_weights.index
    )

    return results


# Conventional 60/40 target weights

weights_60_40 = pd.DataFrame(
    index=monthly_returns.index
)

weights_60_40["SPY"] = 0.60
weights_60_40["IEF"] = 0.40
weights_60_40["BIL"] = 0.00


# Moving-average target weights

weights_6m = create_ma_weights(6)
weights_8m = create_ma_weights(8)
weights_10m = create_ma_weights(10)
weights_12m = create_ma_weights(12)


# Run the net-of-cost backtests

cost_results = {
    "60/40 Portfolio": backtest_with_transaction_costs(
        weights_60_40,
        transaction_cost_bps
    ),

    "6 Month MA": backtest_with_transaction_costs(
        weights_6m,
        transaction_cost_bps
    ),

    "8 Month MA": backtest_with_transaction_costs(
        weights_8m,
        transaction_cost_bps
    ),

    "10 Month MA": backtest_with_transaction_costs(
        weights_10m,
        transaction_cost_bps
    ),

    "12 Month MA": backtest_with_transaction_costs(
        weights_12m,
        transaction_cost_bps
    )
}


# Create a summary table

transaction_cost_summary = {}

for name, result in cost_results.items():

    returns = result["Net Return"]
    values = result["Portfolio Value"]

    years = len(returns) / 12

    annualised_return = (
        values.iloc[-1] / starting_value
    ) ** (1 / years) - 1

    annualised_volatility = (
        returns.std() * (12 ** 0.5)
    )

    drawdown = values / values.cummax() - 1

    annualised_turnover = (
        result["One-Way Turnover"].sum()
        / years
    )

    transaction_cost_summary[name] = {
        "Net Annualised Return (%)":
            annualised_return * 100,

        "Net Annualised Volatility (%)":
            annualised_volatility * 100,

        "Maximum Drawdown (%)":
            drawdown.min() * 100,

        "Annualised Turnover (%)":
            annualised_turnover * 100,

        "Total Transaction Costs ($)":
            result["Transaction Cost ($)"].sum(),

        "Final Value After Costs ($)":
            values.iloc[-1]
    }


transaction_cost_summary = pd.DataFrame(
    transaction_cost_summary
).T

print(
    "Results after",
    transaction_cost_bps,
    "bps transaction costs per dollar bought or sold:"
)

print(transaction_cost_summary.round(2))
```

```python
# 13. Transaction cost sensitivity analysis

cost_levels = [0, 5, 10]

strategy_weights = {
    "60/40 Portfolio": weights_60_40,
    "6 Month MA": weights_6m,
    "8 Month MA": weights_8m,
    "10 Month MA": weights_10m,
    "12 Month MA": weights_12m
}

cost_sensitivity_results = []

for cost_bps in cost_levels:

    for strategy_name, weights in strategy_weights.items():

        result = backtest_with_transaction_costs(
            weights,
            cost_bps
        )

        returns = result["Net Return"]
        values = result["Portfolio Value"]

        years = len(returns) / 12

        annualised_return = (
            values.iloc[-1] / starting_value
        ) ** (1 / years) - 1

        annualised_volatility = (
            returns.std() * (12 ** 0.5)
        )

        drawdown = (
            values / values.cummax() - 1
        )

        cost_sensitivity_results.append({
            "Strategy": strategy_name,
            "Transaction Cost (bps)": cost_bps,
            "Annualised Return (%)": annualised_return * 100,
            "Annualised Volatility (%)": annualised_volatility * 100,
            "Maximum Drawdown (%)": drawdown.min() * 100,
            "Total Transaction Costs ($)": result[
                "Transaction Cost ($)"
            ].sum(),
            "Final Value ($)": values.iloc[-1]
        })


cost_sensitivity_results = pd.DataFrame(
    cost_sensitivity_results
)

print("Transaction cost sensitivity analysis:")
print(
    cost_sensitivity_results.round(2).to_string(index=False)
)
```

```python
# 14. Final portfolio value sensitivity to transaction costs

cost_levels = range(0, 21, 1)

final_value_sensitivity = []

for cost_bps in cost_levels:

    for strategy_name, weights in strategy_weights.items():

        result = backtest_with_transaction_costs(
            weights,
            cost_bps
        )

        final_value_sensitivity.append({
            "Strategy": strategy_name,
            "Transaction Cost (bps)": cost_bps,
            "Final Value ($)": result["Portfolio Value"].iloc[-1]
        })


final_value_sensitivity = pd.DataFrame(
    final_value_sensitivity
)


# Plot final portfolio value against transaction costs

import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

plt.figure(figsize=(11, 6))

for strategy_name in final_value_sensitivity["Strategy"].unique():

    strategy_data = final_value_sensitivity[
        final_value_sensitivity["Strategy"] == strategy_name
    ]

    plt.plot(
        strategy_data["Transaction Cost (bps)"],
        strategy_data["Final Value ($)"],
        marker="o",
        markersize=3,
        label=strategy_name
    )


plt.title(
    "Effect of Transaction Costs on Final Portfolio Value"
)

plt.xlabel(
    "Transaction Cost per Dollar Traded (bps)"
)

plt.ylabel(
    "Final Portfolio Value ($)"
)

plt.gca().yaxis.set_major_formatter(
    FuncFormatter(
        lambda x, position: f"${x:,.0f}"
    )
)

plt.legend()
plt.grid(True)

plt.show()

# 15. Analyse the gradient of transaction cost sensitivity

import numpy as np

gradient_results = []

for strategy_name in final_value_sensitivity["Strategy"].unique():

    strategy_data = final_value_sensitivity[
        final_value_sensitivity["Strategy"] == strategy_name
    ]

    x = strategy_data["Transaction Cost (bps)"]
    y = strategy_data["Final Value ($)"]

    # Fit a straight line: Final Value = gradient * bps + intercept
    gradient, intercept = np.polyfit(x, y, 1)

    # Calculate R-squared to check how linear the relationship is
    predicted_values = gradient * x + intercept

    ss_residual = ((y - predicted_values) ** 2).sum()
    ss_total = ((y - y.mean()) ** 2).sum()

    r_squared = 1 - (ss_residual / ss_total)

    gradient_results.append({
        "Strategy": strategy_name,
        "Gradient ($ per bp)": gradient,
        "Loss per Additional bp ($)": -gradient,
        "Estimated Loss from 10 bps ($)": -gradient * 10,
        "R Squared": r_squared
    })


gradient_results = pd.DataFrame(gradient_results)

# Sort so the most transaction-cost-sensitive strategy appears first
gradient_results = gradient_results.sort_values(
    "Loss per Additional bp ($)",
    ascending=False
)

print("Transaction cost gradient analysis:")
print(
    gradient_results.round(2).to_string(index=False)
)
```

```python
# 17. Overlay value-add and value-destruction attribution

headline_ma_returns = (
    weights_10m["SPY"] * monthly_returns["SPY"]
    + weights_10m["IEF"] * monthly_returns["IEF"]
    + weights_10m["BIL"] * monthly_returns["BIL"]
)

headline_60_40_returns = (
    0.60 * monthly_returns["SPY"]
    + 0.40 * monthly_returns["IEF"]
)

attribution = pd.DataFrame(
    index=headline_ma_returns.index
)

attribution["60/40 Return"] = headline_60_40_returns
attribution["MA Return"] = headline_ma_returns

attribution["Active Return"] = (
    attribution["MA Return"]
    - attribution["60/40 Return"]
)

equity_defensive_signal = (
    weights_10m["SPY"] == 0
).astype(int)

bond_defensive_signal = (
    weights_10m["IEF"] == 0
).astype(int)

attribution["Equity Overlay Contribution"] = (
    equity_defensive_signal
    * 0.60
    * (
        monthly_returns["BIL"]
        - monthly_returns["SPY"]
    )
)

attribution["Bond Overlay Contribution"] = (
    bond_defensive_signal
    * 0.40
    * (
        monthly_returns["BIL"]
        - monthly_returns["IEF"]
    )
)

attribution["Check"] = (
    attribution["Equity Overlay Contribution"]
    + attribution["Bond Overlay Contribution"]
    - attribution["Active Return"]
)


# Add the portfolio state for each month

attribution["Portfolio State"] = headline_states


# Summary by portfolio state

state_summary = attribution.groupby(
    "Portfolio State"
).agg(
    Months=("Active Return", "count"),
    Average_Active_Return=("Active Return", "mean"),
    Total_Active_Return=("Active Return", "sum"),
    Average_MA_Return=("MA Return", "mean"),
    Average_60_40_Return=("60/40 Return", "mean")
)

state_summary[
    [
        "Average_Active_Return",
        "Total_Active_Return",
        "Average_MA_Return",
        "Average_60_40_Return"
    ]
] *= 100

print("\nActive return by portfolio state:")
print(state_summary.round(2))


# Equity-versus-bond attribution

overlay_contribution_summary = pd.Series({
    "Equity Overlay Contribution (%)":
        attribution["Equity Overlay Contribution"].sum() * 100,

    "Bond Overlay Contribution (%)":
        attribution["Bond Overlay Contribution"].sum() * 100,

    "Total Active Return Sum (%)":
        attribution["Active Return"].sum() * 100
})

print("\nOverlay contribution summary:")
print(overlay_contribution_summary.round(2))


# Cumulative relative performance of the 10-month overlay

ma_growth = (
    1 + attribution["MA Return"]
).cumprod()

baseline_growth = (
    1 + attribution["60/40 Return"]
).cumprod()

attribution["Cumulative Relative Performance"] = (
    ma_growth / baseline_growth - 1
)

print(
    "\nFinal cumulative relative performance versus 60/40:",
    round(
        attribution[
            "Cumulative Relative Performance"
        ].iloc[-1] * 100,
        2
    ),
    "%"
)


# Months where the overlay helped most

best_overlay_months = (
    attribution
    .sort_values(
        "Active Return",
        ascending=False
    )
    .head(10)
)

print("\n10 months where the overlay added the most value:")
print(
    (
        best_overlay_months[
            [
                "Portfolio State",
                "60/40 Return",
                "MA Return",
                "Active Return",
                "Equity Overlay Contribution",
                "Bond Overlay Contribution"
            ]
        ]
        .assign(
            **{
                "60/40 Return":
                    best_overlay_months["60/40 Return"] * 100,

                "MA Return":
                    best_overlay_months["MA Return"] * 100,

                "Active Return":
                    best_overlay_months["Active Return"] * 100,

                "Equity Overlay Contribution":
                    best_overlay_months[
                        "Equity Overlay Contribution"
                    ] * 100,

                "Bond Overlay Contribution":
                    best_overlay_months[
                        "Bond Overlay Contribution"
                    ] * 100
            }
        )
        .round(2)
    )
)


# Months where the overlay hurt most

worst_overlay_months = (
    attribution
    .sort_values(
        "Active Return"
    )
    .head(10)
)

print("\n10 months where the overlay destroyed the most value:")
print(
    (
        worst_overlay_months[
            [
                "Portfolio State",
                "60/40 Return",
                "MA Return",
                "Active Return",
                "Equity Overlay Contribution",
                "Bond Overlay Contribution"
            ]
        ]
        .assign(
            **{
                "60/40 Return":
                    worst_overlay_months["60/40 Return"] * 100,

                "MA Return":
                    worst_overlay_months["MA Return"] * 100,

                "Active Return":
                    worst_overlay_months["Active Return"] * 100,

                "Equity Overlay Contribution":
                    worst_overlay_months[
                        "Equity Overlay Contribution"
                    ] * 100,

                "Bond Overlay Contribution":
                    worst_overlay_months[
                        "Bond Overlay Contribution"
                    ] * 100
            }
        )
        .round(2)
    )
)


# Check that the attribution adds up correctly

print(
    "\nLargest attribution calculation error:",
    attribution["Check"].abs().max()
)
```
